```python
import numpy as np
from scipy import stats
from IPython.display import display, Markdown


metricas = {
    "Average time in system": [4.5, 4.395, 4.525, 4.525, 4.28, 4.515, 4.485, 4.43, 4.405, 4.52],
    "Percent idle time": [0.34, 0.35, 0.34, 0.34, 0.34, 0.35, 0.34, 0.35, 0.34, 0.35],
    "Average waiting time per customer": [0.89, 0.8, 0.915, 0.915, 0.655, 0.905, 0.87, 0.865, 0.785, 0.945],
    "Fraction having to wait": [0.37, 0.34, 0.365, 0.38, 0.275, 0.38, 0.38, 0.36, 0.355, 0.4],
    "Average waiting time of those who waited": [2.405405405, 2.352941176, 2.506849315, 2.407894737, 2.381818182, 2.381578947, 2.289473684, 2.402777778, 2.211267606, 2.3625],
}


for nombre, M in metricas.items():

    print("Rep.\tValor")
    for i in range(len(M)):
        print("%d\t%g" % (i + 1, M[i]))

    reps = len(M)
    mean = np.mean(M)
    var = np.var(M, ddof=1)
    desv = np.sqrt(var)

    for alpha in [0.05, 0.01]:
        confidence = (1 - alpha) * 100

        tval = stats.t.ppf(1 - (alpha / 2), reps - 1)
        hval = tval * (desv / np.sqrt(reps))

        display(Markdown(
           rf"""
## {nombre} — {confidence}% confianza

$\bar{{X}} = {mean:.4f}$

$S = {desv:.4f}$

$t_{{n-1,1-\frac{{\alpha}}{2}}} = t_{{{reps-1},{1-alpha/2}}} = {tval:.4f}$

$h = {hval:.4f}$

Intervalo de confianza : $C.I. = (\bar{{X}}-h , \bar{{X}} + h) = ({mean-hval:.4f}, {mean+hval:.4f})$

---
"""))
```
