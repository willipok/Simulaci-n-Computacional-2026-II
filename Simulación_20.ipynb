```python
import numpy as np
from scipy import stats
from IPython.display import display, Markdown


metricas = {
    "Average time in system": [5.05, 4.3, 4.1, 4.05, 4.25, 4.35, 4.55, 4.8, 4.75, 4.55],
    "Percent idle time": [0.24, 0.34, 0.38, 0.35, 0.34, 0.34, 0.36, 0.33, 0.33, 0.34],
    "Average waiting time per customer": [1.5, 0.8, 0.6, 0.5, 0.7, 0.75, 1.05, 1.1, 1.0, 0.85],
    "Fraction having to wait": [0.5, 0.3, 0.25, 0.25, 0.35, 0.33, 0.45, 0.40, 0.40, 0.35],
    "Average waiting time of those who waited": [3.0, 2.66666667, 2.4, 2.0, 2.0, 2.5, 2.33333333, 2.75, 2.5, 2.42857143],
}


for nombre, M in metricas.items():

    print("Rep.\tValor")
    for i in range(len(M)):
        print("%d\t%g" % (i + 1, M[i]))

    reps = len(M)
    mean = np.mean(M)
    var = np.var(M, ddof=1)
    desv = np.sqrt(var)

    for alpha in [0.05, 0.01]:
        confidence = (1 - alpha) * 100

        tval = stats.t.ppf(1 - (alpha / 2), reps - 1)
        hval = tval * (desv / np.sqrt(reps))

        display(Markdown(
           rf"""
## {nombre} — {confidence}% confianza

$\bar{{X}} = {mean:.4f}$

$S = {desv:.4f}$

$t_{{n-1,1-\frac{{\alpha}}{2}}} = t_{{{reps-1},{1-alpha/2}}} = {tval:.4f}$

$h = {hval:.4f}$

Intervalo de confianza : $C.I. = (\bar{{X}}-h , \bar{{X}} + h) = ({mean-hval:.4f}, {mean+hval:.4f})$

---
"""))
```
